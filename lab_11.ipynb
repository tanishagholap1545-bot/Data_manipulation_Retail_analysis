```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("/content/Retail_Dataset_Cleaned (10).csv")
print("Dtaset Loaded Sucessfully")
```

```python
flagship = df[df["Store_Type"] =="Flagship"]["Customer_Satisfaction"]
mall = df[df["Store_Type"]=="Mall Store"]["Customer_Satisfaction"]
```

```python
t_stat, p_value = stats.ttest_ind(
    flagship,
    mall,
    nan_policy="omit"
)
print("T Statistic:", t_stat)
print("P Value:", p_value)
```

```python
if p_value < 0.05:
  print("Reject Null Hypothesis")
else:
  print("Fail to reject Null Hypothesis")
```

```python
df.groupby("Customer_Segment")[
    "Average_Order_Value"].mean()
```

```python
premium = df[
    df["Customer_Segment"]=="Premium"
]["Average_Order_Value"]

budget = df[
    df["Customer_Segment"]=="Budget"
]["Average_Order_Value"]
```

```python
t_stat, p_value = stats.ttest_ind(
    premium,
    budget,
    nan_policy="omit"

)
print("T Statistic:", t_stat)
print("P Value:", p_value)
```

```python
if p_value < 0.05:
  print("Customer segments have significantly different order values.")
else:
  print("No significant difference found.")
```

```python
import matplotlib.pyplot as plt
```

```python
sample_means = []
for i in range (100):
  sample = df["Revenue"].sample(
      n=30,
      replace=True)
  sample_means.append(sample.mean())
```

```python
sample_df = pd.DataFrame({"Sample_Mean": sample_means})
sample_df.head()
```

```python
plt.figure(figsize=(8,5))
plt.hist(sample_df["Sample_Mean"],
         bins=15)
plt.title("Distribution of Revenue Sample Means")
plt.xlabel("Sample Mean Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
print(
    "Mean of Sample Means:",
    sample_df["Sample_Mean"].mean()
)
print(
    "Standard Deviation:",
    sample_df["Sample_Mean"].std()
)
```
