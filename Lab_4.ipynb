```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Retail_Dataset_Modified.csv")
print("Dataset Loaded Successfully")
```

```python
transpose_matrix = sales_matrix.T

print("Transpose Matrix:")
print(transpose_matrix)
```

Reshape

```python
sample_array = np.arange(1, 13)

print(sample_array)
```

```python
matrix_3x4 = sample_array.reshape(3,4)

print(matrix_3x4)
```

```python
matrix_4x3 = sample_array.reshape(4,3)
print(matrix_4x3)
```

```python
flat_array = matrix_3x4.flatten()
print(flat_array)
```
